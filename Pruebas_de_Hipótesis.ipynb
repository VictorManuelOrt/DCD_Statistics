## Introducción

### Ejemplo Simple

Consideremos una situación en la que se efectúa sólo uno de los siguientes dos experimentos aleatorios:

-se lanza un dado equilibrado y se registra el número obtenido

-o bien se lanza una moneda cinco veces y se registra el nímero de cruces totales que se obtienen. (con lados cara y cruz)

![Dado y Moneda](DadoMoneda.PNG)

Ahora, situémonos en la situación en la que únicamente conocemos el resultado reportado x pero no sabemos que experimento se realizó.

Nos planteamos la siguiente pregunta:

### ¿El número x observado es resultado de el experimento del dado o de la monea?

Para resolver esto planteamos dos hipótesis:

$H_0$ : “Se lanzó el dado” vs $H_1$ : “Se lanzó la moneda”.

Antes de intentar resolver este problema vale la pena observar que valores podría tomar x ({0, 1, 2, 3, 4, 5, 6}), y con base en este valor debemos decidir si se llevó a cabo un experimento o el otro.

- Observemos que si el número reportado es 0, entonces con seguridad se realizó el experimento de la moneda (porque el dado no puede marcar 0).
- Si se reporta el número 6, entonces con seguridad el dado fuelanzado (no se podrían habría podido obtener 6 de la moneda porque el dado sólo se lanzó cinco veces).

¿Pero qué hacemos para los otros valores de x?
Una forma de responder esta pregunta es usando máxima verosimilitud. En la siguiente tabla se muestran las probabilidades de obtener los posibles valores de x bajo cada uno de los dos experimentos.

|            | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
|------------|---|---|---|---|---|---|---|
| (D) Dado   | $0$ | >$\frac{1}{6}$<   | $\frac{1}{6}$  |  $\frac{1}{6}$ | >$\frac{1}{6}$<  | >$\frac{1}{6}$<  |  >$\frac{1}{6}$< |
| (M) Moneda | >$\frac{1}{32}$< | $\frac{5}{32}$   |  >$\frac{10}{32}$< |  >$\frac{10}{32}$< | $\frac{5}{32}$  | $\frac{1}{32}$  | $0$  |

Pregunta: ¿Por qué dan estos valores?

Una buena estrategia consiste en decidir por el experimento que tenga mayor probabilidad de producir el valor x observado. Siguiendo esta idea se llega a la siguiente regla de decisión:

### Regla de decisión
### Si $x \in  C = \{0, 2, 3\}$, se rechaza $H_0$ (fue la moneda),
### en caso contrario, no se rechaza $H_0$ (fue el dado).

Al conjunto $C$ se le llama región de rechazo de la hipótesis $H_0$.

Ahora, la regla de decisión anterior no está libre de errores.
- Por ejemplo, si $x = 2$, se decide por el experimento de la moneda, pero el resultado pudo provenir del dado (rechazamos $H_0$ cuando es verdadera).
- Similarmente, si $x = 1$, se decide por el dado pero también pudo ser obtenido por la moneda (no rechazamos $H_0$ cuando es falsa).

Cualquier regla de decisión no estará exenta de errores.

## Tipos de errores en pruebas de hipótesis

Los dos tipos de errores que se pueden presentar se denominan **error tipo I** y **error tipo II**:

|                    | H₀ cierta        | H₀ falsa        |
|--------------------|----------------|----------------|
| **Rechazar H₀**    | Error tipo I   | Decisión correcta |
| **No rechazar H₀** | Decisión correcta | Error tipo II  |


Denotamos por **α** y **β** a las probabilidades de cometer los errores tipo I y II, respectivamente.

Cada uno de estos errores se define y calcula como las siguientes probabilidades condicionales:

- **Error tipo I (α):**  
  $\alpha = P(\text{Rechazar } H_0 \mid H_0 \text{ es verdadera}) $

- **Error tipo II (β):**  
  $ \beta = P(\text{No rechazar } H_0 \mid H_0 \text{ es falsa}) $

### Cálculo de errores tipo I y II (ejemplo dado–moneda)

Sea $X$ la variable aleatoria que representa el número reportado.  
- $D$: evento “lanzar el dado” (hipótesis $H_0$ verdadera)  
- $M$: evento “lanzar la moneda” (hipótesis $H_0$ falsa)

###  Región de rechazo: $C = \{0,2,3\}$

- **Error tipo I:**
  $$
  \alpha = P(\text{“Error tipo I”})= P(\text{“Rechazar H0”}|\text{“H0 es verdadera”}) =P(X \in \{0,2,3\} \mid D) = \frac{2}{6}
  $$

- **Error tipo II:**
  $$
  \beta = P(\text{“Error tipo II”})= P(\text{“No rechazar H0”}|\text{“H0 es falsa}) =P(X \in \{1,4,5,6\} \mid M) = \frac{11}{32}
  $$

> Nota: $\alpha$ y $\beta$ **no suman 1** porque están condicionadas a eventos distintos.

Nos planteamos que ocurre si modificamos la región de rechazo:

---

###  Cambios en la región de rechazo

### Caso 1: $C = \{0,1,2,3\}$

- $$
  \alpha = \frac{3}{6}, \quad \beta = \frac{6}{32}
  $$

 Aumenta $\alpha$, disminuye $\beta$  
 No hay mejor opción sin priorizar errores

---

### Caso 2: $C = \{2,3\}$

- $$
  \alpha = \frac{2}{6}, \quad \beta = \frac{12}{32}
  $$

 $\alpha$ igual, $\beta$ aumenta  
 Se prefiere la región original

---

No siempre se pueden comparar regiones de rechazo directamente ya que reducir un error suele aumentar el otro.

---

###  Criterio general

1. Fijar un nivel máximo para el error tipo I ($\alpha$)
2. Entre todas las regiones que cumplen eso  
    elegir la que **minimiza el error tipo II ($\beta$)**

# Formalización de Conceptos

## Conceptos elementales de pruebas de hipótesis

### Definición: hipótesis estadística
Una **hipótesis estadística** es una afirmación o conjetura acerca de la distribución de una o más variables aleatorias.

En la práctica, suele referirse a afirmaciones sobre parámetros.

---

### Ejemplo
- Si $X \sim \text{Bin}(10, p)$, entonces $p = 0.2$ es una hipótesis.
- Si $X \sim N(\mu, \sigma^2)$, entonces $\mu > 0$ es una hipótesis.

---

### Definición: hipótesis simple y compuesta
- **Hipótesis simple**: especifica completamente la distribución de probabilidad.
- **Hipótesis compuesta**: no especifica completamente la distribución.

---

### Ejemplo
- $X \sim \text{Exp}(\lambda)$, $\lambda = 5$ → hipótesis simple  
- $X \sim N(\mu, 1)$, $\mu = 0$ → hipótesis simple  
- $X \sim \text{Poisson}(\lambda)$, $\lambda > 20$ → hipótesis compuesta  
- $X \sim \chi^2(n)$, $n \neq 5$ → hipótesis compuesta  

---

### Planteamiento del contraste
Se comparan dos hipótesis:

$$
H_0 \; \text{(hipótesis nula)} \quad \text{vs} \quad H_1 \; \text{(hipótesis alternativa)}
$$

Ambas pueden ser simples o compuestas.

---

### Prueba de hipótesis
Consiste en aplicar una regla de decisión para:
- rechazar $H_0$, o
- no rechazar $H_0$

La decisión se basa en una muestra aleatoria.

No rechazar $H_0$ no implica que sea verdadera, solo que es consistente con los datos.

---

### Definición: región de rechazo
Una **región de rechazo** es un subconjunto de valores de la muestra para los cuales se rechaza la hipótesis nula.  
También se llama **región crítica**.

---

### Definición: error tipo I
Se comete cuando se rechaza $H_0$ siendo verdadera.

$$
\alpha = P(\text{rechazar } H_0 \mid H_0 \text{ verdadera})
$$

Se conoce como nivel de significancia.

---

### Definición: error tipo II
Se comete cuando no se rechaza $H_0$ siendo falsa.

$$
\beta = P(\text{no rechazar } H_0 \mid H_0 \text{ falsa})
$$

---

### Observaciones clave
- $\alpha$ y $\beta$ no son complementarios.
- Reducir uno puede aumentar el otro.
- Se busca que ambos sean pequeños.

---

### Estrategia usual
- Fijar un nivel $\alpha$
- Buscar una región de rechazo de tamaño $\alpha$ que minimice $\beta$

Se da prioridad al control del error tipo I.

---


### Resumen de la Aplicación de una Prueba de Hipótesis


- Se define una región de rechazo (por ejemplo, valores extremos de $X$)
- Si el valor observado cae en esa región, se rechaza $H_0$
- Si no, no se rechaza

La elección de la región afecta directamente los valores de $\alpha$ y $\beta$.

---

### Problema central
Encontrar una región de rechazo óptima:
- con tamaño $\alpha$
- que minimice $\beta$

## Importancia de los errores en pruebas de hipótesis

En una prueba de hipótesis, los errores tipo I y tipo II tienen interpretaciones prácticas distintas dependiendo del contexto. Su impacto puede ser muy diferente, por lo que es importante entender cuál error es más costoso.

| Contexto             |$H_0$ vs $H_1$ |Error tipo I (rechazar $H_0$ siendo verdadera) | Error tipo II (no rechazar $H_0$ siendo falsa) |
|---------------------|----|----------------------------------------------|----------------------------------------------|
| Control de calidad  | Producto funcional vs Producto defectuoso |Rechazar un lote bueno (implica pérdidas)                       | Aceptar un lote defectuoso (puede llevar a sanciones)           |
| Diagnóstico médico  | Sano vs Enfermo |Declarar enfermo a alguien sano (gastos médicos innecesarios)     | No detectar una enfermedad (complicación médica)             |

---

### Interpretación

- **Error tipo I ($\alpha$)**:
  - Puede generar costos innecesarios o decisiones injustificadas.
  - Ejemplo: desechar productos buenos o alarmar a una persona sana.

- **Error tipo II ($\beta$)**:
  - Puede ser más grave en contextos críticos.
  - Ejemplo: permitir productos defectuosos o no detectar una enfermedad.

---

### Conclusión

La elección del nivel de significancia $\alpha$ depende del contexto:

- Si el **error tipo I es más grave**, se controla estrictamente $\alpha$.
- Si el **error tipo II es más grave**, se busca reducir $\beta$.

No siempre es posible minimizar ambos errores al mismo tiempo, por lo que se debe priorizar según el problema.

## Ejercicio:
Sea $X_1$, $X_2$, $X_3$ una muestra aleatoria  de la distribución $N(\theta, 4)$, en donde $\theta$ es desconocido. Encuentre las probabilidades de cometer los errores tipo I y II para la prueba $$H_0 : \theta = 2 \text{ vs } H_1 : \theta  = 5$$,
considerando como región de rechazo:
- a) $C = \{ (x_1, x_2, x_3 : x_1 > 4.7\}$.
- b) $C = \{ (x_1, x_2, x_3 : (x_1 + 2x_2)/3 >4.5\}$.
- c) $C = \{ (x_1, x_2, x_3 :(x_1 + x_3)/2 >4.2\}$.
- d) $C = \{ (x_1, x_2, x_3 : \overline{x} > 4.1\}$.

## Cálculo de $\alpha$ y $\beta$ (incisos a y c)

Recordemos:
- $X_i \sim N(\theta, 4)$ ⇒ $\sigma = 2$
- $\alpha = P(\text{rechazar } H_0 \mid \theta = 2)$
- $\beta = P(\text{no rechazar } H_0 \mid \theta = 5)$

Usaremos estandarización:
$$
Z = \frac{X - \mu}{\sigma}
$$

---

## a) $C = \{x_1 > 4.7\}$

Aquí:
$$
X_1 \sim N(\theta, 4)
$$

### Error tipo I ($\alpha$)
Bajo $H_0: \theta = 2$:
$$
\alpha = P(X_1 > 4.7 \mid \theta = 2)
$$

Estandarizamos:
$$
z = \frac{4.7 - 2}{2} = \frac{2.7}{2} = 1.35
$$

Entonces:
$$
\alpha = P(Z > 1.35)
$$

De tablas:
$$
\alpha \approx 0.0885
$$

---

### Error tipo II ($\beta$)
Bajo $H_1: \theta = 5$:
$$
\beta = P(X_1 \leq 4.7 \mid \theta = 5)
$$

Estandarizamos:
$$
z = \frac{4.7 - 5}{2} = \frac{-0.3}{2} = -0.15
$$

Entonces:
$$
\beta = P(Z \leq -0.15)
$$

De tablas:
$$
\beta \approx 0.4404
$$

---

## c) $C = \left\{ \frac{X_1 + X_3}{2} > 4.2 \right\}$

Definimos:
$$
Y = \frac{X_1 + X_3}{2}
$$

### Distribución de $Y$

Como $X_1, X_3$ son independientes:
- $E[Y] =\frac{E[X_1]+E[X_3]}{2}=\frac{\theta+\theta}{2}= \theta$
- $Var(Y) = \frac{1}{4}(Var(X_1) + Var(X_3)) = \frac{1}{4}(4 + 4) = 2$

Entonces:
$$
Y \sim N(\theta, 2), \quad \sigma_Y = \sqrt{2}
$$

---

### Error tipo I ($\alpha$)
Bajo $H_0: \theta = 2$:
$$
\alpha = P(Y > 4.2)
$$

Estandarizamos:
$$
z = \frac{4.2 - 2}{\sqrt{2}} = \frac{2.2}{\sqrt{2}} \approx 1.56
$$

Entonces:
$$
\alpha = P(Z > 1.56)
$$

De tablas:
$$
\alpha \approx 0.0599
$$

---

### Error tipo II ($\beta$)
Bajo $H_1: \theta = 5$:
$$
\beta = P(Y \leq 4.2)
$$

Estandarizamos:
$$
z = \frac{4.2 - 5}{\sqrt{2}} = \frac{-0.8}{\sqrt{2}} \approx -0.57
$$

Entonces:
$$
\beta = P(Z \leq -0.57)
$$

De tablas:
$$
\beta \approx 0.2858
$$

````python
import math
from scipy.stats import norm

# Parámetros
mu0 = 2   # bajo H0
mu1 = 5   # bajo H1
sigma = 2 # desviación estándar (varianza = 4)

# Funciones auxiliares
def alpha(threshold, mean, std):
    return 1 - norm.cdf(threshold, loc=mean, scale=std)

def beta(threshold, mean, std):
    return norm.cdf(threshold, loc=mean, scale=std)

# a) X1 > 4.7
std_a = sigma
alpha_a = alpha(4.7, mu0, std_a)
beta_a = beta(4.7, mu1, std_a)

# b) (X1 + 2X2)/3 > 4.5
std_b = math.sqrt(20/9)
alpha_b = alpha(4.5, mu0, std_b)
beta_b = beta(4.5, mu1, std_b)

# c) (X1 + X3)/2 > 4.2
std_c = math.sqrt(2)
alpha_c = alpha(4.2, mu0, std_c)
beta_c = beta(4.2, mu1, std_c)

# d) X̄ > 4.1
std_d = math.sqrt(4/3)
alpha_d = alpha(4.1, mu0, std_d)
beta_d = beta(4.1, mu1, std_d)

# Mostrar resultados
print("a) alpha =", alpha_a, "beta =", beta_a)
print("b) alpha =", alpha_b, "beta =", beta_b)
print("c) alpha =", alpha_c, "beta =", beta_c)
print("d) alpha =", alpha_d, "beta =", beta_d)
````

## Ejemplo: detección de contenedores defectuosos

En una fábrica, los paquetes de café tienen peso:
- $X \sim N(500, 50)$ (proceso correcto)
- $X \sim N(490, 50)$ (proceso defectuoso)

Se toma una muestra de tamaño 2: $(X_1, X_2)$.

---

### Planteamiento de hipótesis

$$
H_0: X \sim N(500, 50) \\
H_1: X \sim N(490, 50)
$$

---

### Región de rechazo (región crítica)

Se acepta $H_1$ si:

$$
G = \{(x_1, x_2): x_1 \leq 496 \text{ y } x_2 \leq 496\}
$$

Es decir, **si ambos valores son pequeños**, se considera que el lote es defectuoso.

---

## Cálculo de errores

### Error tipo I ($\alpha$)

Rechazar $H_0$ cuando es verdadera:

$$
\alpha = P(X_1 \leq 496, X_2 \leq 496 \mid \mu = 500)
$$

Como $X_1, X_2$ son independientes:

$$
\alpha = \left[P(X \leq 496 \mid \mu = 500)\right]^2
$$

Estandarizando:

$$
z = \frac{496 - 500}{\sqrt{50}} \approx -0.566
$$

Entonces:

$$
P(X \leq 496) = \Phi(-0.566) \approx 0.286
$$

$$
\alpha \approx (0.286)^2 \approx 0.0817
$$

---

### Error tipo II ($\beta$)

No rechazar $H_0$ cuando es falsa:

$$
\beta = P(X_1> 496 \text{ ó } X_2 > 496 \mid \mu = 490)
$$

$$
\beta = 1 - P(X_1 \leq 496, X_2 \leq 496 \mid \mu = 490)
$$

$$
\beta = 1 - \left[P(X \leq 496 \mid \mu = 490)\right]^2
$$

Estandarizando:

$$
z = \frac{496 - 490}{\sqrt{50}} \approx 0.849
$$

$$
P(X \leq 496) = \Phi(0.849) \approx 0.802
$$

$$
\beta \approx 1 - (0.802)^2 \approx 0.3569
$$

---

## Código en Python

```python
import math
from scipy.stats import norm

sigma = math.sqrt(50)

# Error tipo I
alpha = (norm.cdf((496 - 500)/sigma))**2

# Error tipo II
beta = 1 - (norm.cdf((496 - 490)/sigma))**2

print("alpha =", alpha)
print("beta =", beta)

## ¿Qué es el estadístico de prueba?

El **estadístico de prueba** es una función de la muestra aleatoria:

$$
T = T(X_1, X_2, \dots, X_n)
$$

que se utiliza para:
- resumir la información de los datos
- tomar la decisión de rechazar o no $H_0$

---

## Interpretación

Es un número que:
- depende de los datos observados
- tiene una **distribución conocida bajo $H_0$**
- permite comparar contra una región crítica o un valor crítico

---

## Ejemplo típico

$$
T = \frac{\bar{X} - \mu_0}{\sigma / \sqrt{n}}
$$

- Bajo $H_0$, $T \sim N(0,1)$
- Se usa para decidir si $\bar{X}$ es “demasiado grande o pequeño”

---

## ¿Dónde aparece en el proceso?

Orden lógico de una prueba de hipótesis:

### 1. Plantear hipótesis
$$
H_0 \quad \text{vs} \quad H_1
$$

---

### 2. Definir el estadístico de prueba
- Elegir una función $T(X)$
- Conocer su distribución bajo $H_0$

---

### 3. Definir la región de rechazo
- En términos de $T$:
$$
\{T \geq c\}, \quad \{T \leq c\}, \quad \text{o} \quad \{|T| \geq c\}
$$

---

### 4. Fijar $\alpha$ y obtener el valor crítico
- Determinar $c$

---

### 5. Calcular el valor observado
$$
t_{\text{obs}} = T(x_1, \dots, x_n)
$$

---

### 6. Tomar decisión
- Si $t_{\text{obs}}$ cae en la región crítica → rechazar $H_0$
- Si no → no rechazar


## Ejemplo de estadístico de prueba

Supongamos que queremos verificar si el peso promedio de un producto es menor a 500 g.

---

### 1. Hipótesis

$$
H_0: \mu = 500 \quad \text{vs} \quad H_1: \mu < 500
$$

---

### 2. Modelo

- $X_1, \dots, X_n \sim N(\mu, \sigma^2)$
- $\sigma$ conocida

---

### 3. Estadístico de prueba

Definimos:

$$
T = \frac{\bar{X} - 500}{\sigma / \sqrt{n}}
$$

Bajo $H_0$:

$$
T \sim N(0,1)
$$

---

### 4. Región de rechazo

Como es prueba de cola izquierda:

$$
\text{Rechazar } H_0 \text{ si } T \leq z_\alpha
$$

Por ejemplo, si $\alpha = 0.05$:

$$
z_\alpha \approx -1.645
$$

---

### 5. Datos observados

Supongamos:
- $\bar{x} = 495$
- $\sigma = 10$
- $n = 25$

Entonces:

$$
T = \frac{495 - 500}{10 / \sqrt{25}} = \frac{-5}{2} = -2.5
$$

---

### 6. Decisión

- $T = -2.5$
- $z_\alpha = -1.645$

Como:

$$
-2.5 < -1.645
$$

 **Rechazamos $H_0$**

---

## Interpretación

Hay evidencia suficiente para concluir que:
- el peso promedio es menor a 500 g


## Lema de Neyman–Pearson

### Planteamiento

Sea una muestra $X$ con función de verosimilitud $L(X, \theta )$.

Queremos contrastar dos hipótesis **simples**:

$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta = \theta_1
$$

---

### Enunciado del lema

Entre todas las pruebas con nivel de significancia $\alpha$, la prueba más potente (es decir, la que minimiza $\beta$) es aquella cuya región de rechazo tiene la forma:

$$
C = \left\{ x : \frac{L(x, \theta_1 )}{L(x, \theta_0 )} > k \right\}
$$

donde $k$ es una constante elegida para que:

$$
P(X \in C \mid \theta = \theta_0) = \alpha
$$

---

### Interpretación

- Se compara qué tan probable es observar los datos bajo $H_1$ vs $H_0$
- Si los datos son **mucho más probables bajo $H_1$**, se rechaza $H_0$

El cociente:

$$
\frac{L(x, \theta_1)}{L(x, \theta_0 )}
$$

se llama **razón de verosimilitudes**.

---

### Idea intuitiva

- Si los datos favorecen más a $H_1$ que a $H_0$, se rechaza $H_0$
- La región crítica contiene los valores donde:
  - $H_1$ explica mejor los datos que $H_0$

---

### Qué significa "más potente"

La potencia es:

$$
1 - \beta = P(\text{rechazar } H_0 \mid H_1 \text{ verdadera})
$$

Entonces el lema dice:

> Ninguna otra prueba con el mismo $\alpha$ tiene mayor probabilidad de detectar que $H_0$ es falsa.

---

### Importancia

- Da la **mejor prueba posible** en el caso simple vs simple
- Justifica por qué usamos razones de verosimilitud

---


## Ejemplo: prueba óptima para una normal con varianza conocida

Sea $X \sim N(\theta, \sigma^2)$, donde $\theta$ es desconocido y $\sigma^2$ es conocida.

Se desea contrastar las hipótesis simples:

$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta = \theta_1
$$

donde $\theta_0 < \theta_1$.

Se tiene una muestra aleatoria $X_1, \dots, X_n$.

---

## Función de verosimilitud

$$
L(x^n, \theta) = \left( \frac{1}{2\pi\sigma^2} \right)^{n/2}
\exp\left( -\frac{1}{2\sigma^2} \sum_{i=1}^n (x_i - \theta)^2 \right)
$$

---

## Cociente de verosimilitudes

$$
\frac{L(x^n, \theta_1)}{L(x^n, \theta_0)} =
\exp\left(
-\frac{1}{2\sigma^2}
\sum_{i=1}^n \left[(x_i - \theta_1)^2 - (x_i - \theta_0)^2 \right]
\right)
$$

Simplificando:

$$
= \exp\left(
-\frac{1}{2\sigma^2}
\left[
n(\theta_1^2 - \theta_0^2) - 2n\bar{x}(\theta_1 - \theta_0)
\right]
\right)
$$

---

## Región de rechazo

Por el lema de Neyman–Pearson, se rechaza $H_0$ si:

$$
\frac{L(\theta_1 \mid x)}{L(\theta_0 \mid x)} \geq c
$$

Esto es equivalente a:

$$
\bar{x} \geq c
$$

para alguna constante $c$.

Entonces, la región crítica es:

$$
C = \{(x_1, \dots, x_n): \bar{x} \geq c\}
$$

---

## Cálculo de $c$ (nivel de significancia)

Queremos:

$$
\alpha = P(\bar{X} \geq c \mid \theta = \theta_0)
$$

Estandarizando:

$$
\alpha = P\left(
\frac{\bar{X} - \theta_0}{\sigma/\sqrt{n}} \geq
\frac{c - \theta_0}{\sigma/\sqrt{n}}
\right)
= 1 - \Phi\left( \frac{c - \theta_0}{\sigma/\sqrt{n}} \right)
$$

Despejando $c$:

$$
c = \theta_0 + \frac{\sigma}{\sqrt{n}} \Phi^{-1}(1 - \alpha)
$$

---

## Probabilidad de error tipo II

$$
\beta = P(\bar{X} < c \mid \theta = \theta_1)
$$

Estandarizando:

$$
\beta = \Phi\left( \frac{c - \theta_1}{\sigma/\sqrt{n}} \right)
$$

Sustituyendo $c$:

$$
\beta = \Phi\left(
\frac{\theta_0 - \theta_1}{\sigma/\sqrt{n}} + \Phi^{-1}(1 - \alpha)
\right)
$$


## EJERCICIO: PYTHON
• Un investigador desea analizar si los pasajeros de primera clase del Titanic eran, en promedio, más jóvenes que el total de pasajeros del barco. Para ello, se cuenta con la base de datos titanic disponible en la biblioteca seaborn de Python, que contiene, entre otras variables, la edad (age) y la clase del pasajero (pclass). Se sabe que la media general de edad de todos los pasajeros es $\mu_\text{total}$ ,y se desea contrastar si la media de edad de los pasajeros de primera clase $\mu_\text{primera}$ es menor que dicha media poblacional. (Agregar gráficos de la región crítica)

$H_0$: La media de primera clase es mayor o igual a la media poblacional

$H_1$: La media de primera clase es menor a la media poblacional

Haremos una prueba de cola izquierda
Porque estamos buscando valores menores, pues mientras más pequeño sea nuestro estadístico de prueba, más evidencia contra $H_0$.	​


````python
import seaborn as sns
import numpy as np
import math
import matplotlib.pyplot as plt
from scipy.stats import norm

# Cargar y limpiar datos
df = sns.load_dataset("titanic").dropna(subset=["age"])

# Estadísticos
mu_total = df["age"].mean()
first_class = df[df["pclass"] == 1]["age"]

xbar = first_class.mean()
n = len(first_class)
s = first_class.std(ddof=1)

# Hipótesis
alpha = 0.05

z_stat = (xbar - mu_total) / (s / math.sqrt(n))
z_crit = norm.ppf(alpha)

# Rango enfocado
x = np.linspace(-4, 10, 500)
y = norm.pdf(x)

plt.figure(figsize=(8,5))

# Curva normal
plt.plot(x, y)

# Región crítica sombreada
plt.fill_between(x, y, where=(x <= z_crit), alpha=0.4)

# Líneas verticales
plt.axvline(z_crit, linestyle='--')
plt.axvline(z_stat, linestyle='--')

# Etiquetas claras
plt.text(z_crit, max(y)*0.9, f"z crítico = {z_crit:.2f}", rotation=90)
plt.text(z_stat, max(y)*0.6, f"z observado = {z_stat:.2f}", rotation=90)

# Títulos
plt.title("Prueba de hipótesis (cola izquierda)")
plt.xlabel("Estadístico Z")
plt.ylabel("Densidad")

plt.show()

# Decisión automática
if z_stat < z_crit:
    print("Rechazar H0")
else:
    print("No rechazar H0")
````

## p-valor

### Definición
El **p-valor** es la probabilidad de observar un resultado tan extremo como el obtenido (o más), **suponiendo que $H_0$ es verdadera**.

---

### Interpretación

Responde a la pregunta:

> Si $H_0$ fuera cierta, ¿qué tan raro es el resultado observado?

- **p-valor pequeño** → evidencia contra $H_0$
- **p-valor grande** → datos compatibles con $H_0$

---

### Regla de decisión

Dado un nivel de significancia $\alpha$:

- Si $p\text{-valor} \leq \alpha$ → **rechazar $H_0$**
- Si $p\text{-valor} > \alpha$ → **no rechazar $H_0$**

---

### Cálculo del p-valor

Depende del tipo de prueba:

#### Cola izquierda ($H_1: \mu < \mu_0$)

$$
p\text{-valor} = P(Z \leq z_{\text{obs}})
$$

---

#### Cola derecha ($H_1: \mu > \mu_0$)

$$
p\text{-valor} = P(Z \geq z_{\text{obs}})
$$

---

#### Dos colas ($H_1: \mu \neq \mu_0$)

$$
p\text{-valor} = 2 \cdot P(Z \geq |z_{\text{obs}}|)
$$

---

### Ejemplo

Si:
$$
z_{\text{obs}} = -2.1
$$

y la prueba es de cola izquierda:

$$
p\text{-valor} = P(Z \leq -2.1) \approx 0.018
$$

---

### Interpretación del ejemplo

- $p = 0.018$
- Si $\alpha = 0.05$:

$$
0.018 < 0.05 \Rightarrow \text{rechazar } H_0
$$

---

### Error común

El p-valor **no es**:
- la probabilidad de que $H_0$ sea verdadera
- la probabilidad de cometer un error

Es una probabilidad calculada **bajo el supuesto de que $H_0$ es verdadera**.



## Resumen

En esta sección se estudian pruebas de hipótesis para parámetros de la distribución normal.

---

### Definición de $z_\alpha$

Se define $z_\alpha$ como el número real que satisface:

$$
P(Z > z_\alpha) = \alpha
$$

donde:
- $\alpha \in (0,1)$
- $Z \sim N(0,1)$ (normal estándar)

---

### Interpretación

- $z_\alpha$ es el valor que deja una probabilidad $\alpha$ **a la derecha**
- Es un valor crítico usado en pruebas de hipótesis

---

### Función de distribución

La función de distribución de la normal estándar es:

$$
\Phi(x) = P(Z \leq x)
$$

---

### Relación importante

$$
\Phi(z_\alpha) = 1 - \alpha
$$

---

### Idea clave

- $z_\alpha$ define regiones críticas en pruebas de hipótesis
- Se usa para controlar el nivel de significancia $\alpha$

## Pruebas para la media con varianza conocida

Sea $X_1, \dots, X_n \sim N(\theta, \sigma^2)$, con:
- $\theta$ desconocido
- $\sigma^2$ conocida

Entonces:

$$
\bar{X} \sim N\left(\theta, \frac{\sigma^2}{n}\right)
$$

y

$$
Z_0 = \frac{\bar{X} - \theta}{\sigma / \sqrt{n}} \sim N(0,1)
$$

---

## Estadístico de prueba

Para contrastar un valor $\theta_0$:

$$
Z_0 = \frac{\bar{X} - \theta_0}{\sigma / \sqrt{n}}
$$

Este mide qué tan lejos está $\bar{X}$ de $\theta_0$ bajo $H_0$.

---

## Prueba de dos colas

### Hipótesis

$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta \neq \theta_0
$$

### Región de rechazo

$$
|Z_0| \geq z_{\alpha/2}
$$

equivalentemente:

$$
C = \left\{ (\mathbf{x}) : \left| \frac{\bar{x} - \theta_0}{\sigma / \sqrt{n}} \right| \geq z_{\alpha/2} \right\}
$$

### Error tipo II

$$
\beta(\theta_1) =
\Phi\left(z_{\alpha/2} + \frac{\theta_0 - \theta_1}{\sigma/\sqrt{n}}\right)
-
\Phi\left(-z_{\alpha/2} + \frac{\theta_0 - \theta_1}{\sigma/\sqrt{n}}\right)
$$

---

## Prueba de cola izquierda

### Hipótesis

$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta < \theta_0
$$

### Región de rechazo

$$
Z_0 \leq -z_\alpha
$$

equivalentemente:

$$
C = \left\{ (\mathbf{x}) :
\frac{\bar{x} - \theta_0}{\sigma / \sqrt{n}} \leq -z_\alpha
\right\}
$$

### Error tipo II

$$
\beta(\theta_1) =
1 - \Phi\left(-z_\alpha + \frac{\theta_0 - \theta_1}{\sigma/\sqrt{n}}\right)
$$

---

## Prueba de cola derecha

### Hipótesis

$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta > \theta_0
$$

### Región de rechazo

$$
Z_0 \geq z_\alpha
$$

equivalentemente:

$$
C = \left\{ (\mathbf{x}) :
\frac{\bar{x} - \theta_0}{\sigma / \sqrt{n}} \geq z_\alpha
\right\}
$$

### Error tipo II

$$
\beta(\theta_1) =
\Phi\left(z_\alpha + \frac{\theta_0 - \theta_1}{\sigma/\sqrt{n}}\right)
$$

---

## Resumen

- Todas las pruebas se basan en el mismo estadístico $Z_0$
- La diferencia está en la **región crítica**
- Se fija $\alpha$  y se obtiene $z_\alpha$ o $z_{\alpha/2}$
- Luego se decide según dónde cae $Z_0$

---

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

alpha = 0.05

# Valores críticos
z_alpha = norm.ppf(1 - alpha)
z_alpha_2 = norm.ppf(1 - alpha/2)

x = np.linspace(-4, 4, 500)
y = norm.pdf(x)

# -------------------------------
# Prueba de dos colas
# -------------------------------
plt.figure()
plt.plot(x, y)

# Regiones de rechazo
plt.fill_between(x, y, where=(x <= -z_alpha_2))
plt.fill_between(x, y, where=(x >= z_alpha_2))

# Líneas críticas
plt.axvline(-z_alpha_2)
plt.axvline(z_alpha_2)

plt.title("Región de rechazo (dos colas)")
plt.xlabel("Z")
plt.ylabel("Densidad")
plt.show()


# -------------------------------
# Cola izquierda
# -------------------------------
plt.figure()
plt.plot(x, y)

plt.fill_between(x, y, where=(x <= -z_alpha))
plt.axvline(-z_alpha)

plt.title("Región de rechazo (cola izquierda)")
plt.xlabel("Z")
plt.ylabel("Densidad")
plt.show()


# -------------------------------
# Cola derecha
# -------------------------------
plt.figure()
plt.plot(x, y)

plt.fill_between(x, y, where=(x >= z_alpha))
plt.axvline(z_alpha)

plt.title("Región de rechazo (cola derecha)")
plt.xlabel("Z")
plt.ylabel("Densidad")
plt.show()
````

## Pruebas para la media con varianza desconocida

Sea $X_1, \dots, X_n \sim N(\theta, \sigma^2)$, donde:
- $\theta$ es desconocido
- $\sigma^2$ es **desconocida**

---

## Resultado clave

Se utiliza la varianza muestral $S^2$ y se tiene:

$$
T_0 = \frac{\bar{X} - \theta}{S / \sqrt{n}} \sim t(n-1)
$$

donde $t(n-1)$ es la distribución t de Student con $n-1$ grados de libertad.

---

## Estadístico de prueba

Para contrastar $\theta_0$:

$$
T_0 = \frac{\bar{X} - \theta_0}{S / \sqrt{n}}
$$

---

## Prueba de dos colas

### Hipótesis

$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta \neq \theta_0
$$

### Región de rechazo

$$
|T_0| \geq t_{\alpha/2,\, n-1}
$$

equivalentemente:

$$
C = \left\{ (\mathbf{x}) :
\left| \frac{\bar{x} - \theta_0}{s / \sqrt{n}} \right| \geq t_{\alpha/2,\, n-1}
\right\}
$$

### Error tipo II (aproximado)

$$
\beta(\theta_1) \approx
F\left(t_{\alpha/2,\, n-1} + \frac{\theta_0 - \theta_1}{s/\sqrt{n}}\right)
-
F\left(-t_{\alpha/2,\, n-1} + \frac{\theta_0 - \theta_1}{s/\sqrt{n}}\right)
$$

---

## Prueba de cola izquierda

### Hipótesis

$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta < \theta_0
$$

### Región de rechazo

$$
T_0 \leq t_{\alpha,\, n-1}
$$

equivalentemente:

$$
C = \left\{ (\mathbf{x}) :
\frac{\bar{x} - \theta_0}{s / \sqrt{n}} \leq t_{\alpha,\, n-1}
\right\}
$$

### Error tipo II (aproximado)

$$
\beta(\theta_1) \approx
1 - F\left(t_{\alpha,\, n-1} + \frac{\theta_0 - \theta_1}{s/\sqrt{n}}\right)
$$

---

## Prueba de cola derecha

### Hipótesis

$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta > \theta_0
$$

### Región de rechazo

$$
T_0 \geq t_{\alpha,\, n-1}
$$

equivalentemente:

$$
C = \left\{ (\mathbf{x}) :
\frac{\bar{x} - \theta_0}{s / \sqrt{n}} \geq t_{\alpha,\, n-1}
\right\}
$$

### Error tipo II (aproximado)

$$
\beta(\theta_1) \approx
F\left(t_{\alpha,\, n-1} + \frac{\theta_0 - \theta_1}{s/\sqrt{n}}\right)
$$

---

## Resumen

- Cuando $\sigma^2$ es desconocida, se usa la **distribución t de Student**
- Se reemplaza $\sigma$ por $S$
- Las regiones críticas dependen de cuantiles $t_{\alpha, n-1}$



````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t

# Parámetros
alpha = 0.05
n = 20                  # tamaño de muestra
gl = n - 1              # grados de libertad

# Valores críticos
t_alpha = t.ppf(1 - alpha, gl)
t_alpha_2 = t.ppf(1 - alpha/2, gl)

# Rango
x = np.linspace(-4, 4, 500)
y = t.pdf(x, gl)

# -------------------------------
# Prueba de dos colas
# -------------------------------
plt.figure()
plt.plot(x, y)

plt.fill_between(x, y, where=(x <= -t_alpha_2))
plt.fill_between(x, y, where=(x >= t_alpha_2))

plt.axvline(-t_alpha_2)
plt.axvline(t_alpha_2)

plt.title("t-Student: Región de rechazo (dos colas)")
plt.xlabel("T")
plt.ylabel("Densidad")
plt.show()


# -------------------------------
# Cola izquierda
# -------------------------------
plt.figure()
plt.plot(x, y)

plt.fill_between(x, y, where=(x <= -t_alpha))
plt.axvline(-t_alpha)

plt.title("t-Student: Región de rechazo (cola izquierda)")
plt.xlabel("T")
plt.ylabel("Densidad")
plt.show()


# -------------------------------
# Cola derecha
# -------------------------------
plt.figure()
plt.plot(x, y)

plt.fill_between(x, y, where=(x >= t_alpha))
plt.axvline(t_alpha)

plt.title("t-Student: Región de rechazo (cola derecha)")
plt.xlabel("T")
plt.ylabel("Densidad")
plt.show()
````

H_0: \sigma^2 = \sigma_0^2

La estadística:

$$
\chi_0^2 = \frac{(n-1)S^2}{\sigma_0^2} \sim \chi^2(n-1)
$$

donde $S^2$ es la varianza muestral.

- Valor esperado: $E(\chi_0^2) = n-1$  
- Idea: rechazar $H_0$ cuando $\chi_0^2$ esté lejos de $n-1$

---

## Prueba de dos colas

### Hipótesis

$$
H_0: \sigma^2 = \sigma_0^2 \quad \text{vs} \quad H_1: \sigma^2 \neq \sigma_0^2
$$

### Región de rechazo

$$
\chi_0^2 < \chi^2_{1-\alpha/2} \quad \text{o} \quad \chi_0^2 > \chi^2_{\alpha/2}
$$

---

### Error tipo II

$$
\beta(\sigma_1^2) =
F\left(\chi^2_{\alpha/2} \cdot \frac{\sigma_0^2}{\sigma_1^2}\right)
-
F\left(\chi^2_{1-\alpha/2} \cdot \frac{\sigma_0^2}{\sigma_1^2}\right)
$$

---

## Prueba de cola izquierda

### Hipótesis

$$
H_0: \sigma^2 = \sigma_0^2 \quad \text{vs} \quad H_1: \sigma^2 < \sigma_0^2
$$

### Región de rechazo

$$
\chi_0^2 < \chi^2_{1-\alpha}
$$

---

### Error tipo II

$$
\beta(\sigma_1^2) =
1 - F\left(\chi^2_{1-\alpha} \cdot \frac{\sigma_0^2}{\sigma_1^2}\right)
$$

---

## Prueba de cola derecha

### Hipótesis

$$
H_0: \sigma^2 = \sigma_0^2 \quad \text{vs} \quad H_1: \sigma^2 > \sigma_0^2
$$

### Región de rechazo

$$
\chi_0^2 > \chi^2_{\alpha}
$$

---

### Error tipo II

$$
\beta(\sigma_1^2) =
F\left(\chi^2_{\alpha} \cdot \frac{\sigma_0^2}{\sigma_1^2}\right)
$$

---


````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

# Parámetros
alpha = 0.05
n = 20
gl = n - 1  # grados de libertad

# Valores críticos
chi_left = chi2.ppf(alpha/2, gl)
chi_right = chi2.ppf(1 - alpha/2, gl)

chi_left_one = chi2.ppf(alpha, gl)
chi_right_one = chi2.ppf(1 - alpha, gl)

# Rango
x = np.linspace(0, 50, 500)
y = chi2.pdf(x, gl)

# -------------------------------
# Prueba de dos colas
# -------------------------------
plt.figure()
plt.plot(x, y)

plt.fill_between(x, y, where=(x <= chi_left))
plt.fill_between(x, y, where=(x >= chi_right))

plt.axvline(chi_left)
plt.axvline(chi_right)

plt.title("Región de rechazo (dos colas) - Chi-cuadrado")
plt.xlabel("χ²")
plt.ylabel("Densidad")
plt.show()


# -------------------------------
# Cola izquierda
# -------------------------------
plt.figure()
plt.plot(x, y)

plt.fill_between(x, y, where=(x <= chi_left_one))
plt.axvline(chi_left_one)

plt.title("Región de rechazo (cola izquierda) - Chi-cuadrado")
plt.xlabel("χ²")
plt.ylabel("Densidad")
plt.show()


# -------------------------------
# Cola derecha
# -------------------------------
plt.figure()
plt.plot(x, y)

plt.fill_between(x, y, where=(x >= chi_right_one))
plt.axvline(chi_right_one)

plt.title("Región de rechazo (cola derecha) - Chi-cuadrado")
plt.xlabel("χ²")
plt.ylabel("Densidad")
plt.show()
````

## Ejemplos

En ciertas zonas de la ciudad y durante varios periodos anuales se ha calculado
el pago por el consumo de agua suponiendo un consumo promedio de
20,000 litros mensuales en cada casa. Para determinar si tal cantidad
ha cambiado, se han medido los consumos mensuales de 15 casas escogidas al azar, obteniendose los siguientes resultados: 23456, 18325,
21982, 22371, 13292, 25073, 22601, 20930, 18788, 19162, 21442, 23935,
20320, 19095, 17421. ¿Debe cambiar el consumo promedio mensual
estimado para el calculo de los pagos o permanecer igual? Suponga
$\sigma=$2000.

$H_0​:\mu=20000\text{ vs }H1_​:\mu\neq20000$

````python
import numpy as np
from scipy.stats import norm

# Datos
data = np.array([
    23456, 18325, 21982, 22371, 13292,
    25073, 22601, 20930, 18788, 19162,
    21442, 23935, 20320, 19095, 17421
])

# Parámetros
mu0 = 20000
sigma = 2000
n = len(data)

# Estadísticos
xbar = np.mean(data)

# Estadístico Z
z = (xbar - mu0) / (sigma / np.sqrt(n))

# Valor crítico (dos colas)
alpha = 0.05
z_crit = norm.ppf(1 - alpha/2)

# p-valor
p_value = 2 * (1 - norm.cdf(abs(z)))

print("Media muestral:", xbar)
print("Z:", z)
print("z crítico:", z_crit)
print("p-valor:", p_value)

# Decisión
if abs(z) >= z_crit:
    print("Se rechaza H0")
else:
    print("No se rechaza H0")
````

En una muestra aleatoria, el tiempo promedio en el que 50 mujeres
terminaron una prueba escrita fue de 30 minutos, mientras que 45
hombres terminaron la prueba en un promedio de 35 minutos. Para
fines ilustrativos supondremos una varianza de 9 unidades en ambas
poblaciones. ¿Hay alguna diferencia entre hombres y mujeres en el
tiempo promedio real para concluir la prueba?

$H_0​:\mu_M​=\mu_H​$

````python
import numpy as np
from scipy.stats import norm

# Datos
xbar_M = 30
xbar_H = 35

n_M = 50
n_H = 45

sigma = 3  # desviación estándar (raíz de 9)

# Estadístico Z
z = (xbar_M - xbar_H) / np.sqrt((sigma**2)/n_M + (sigma**2)/n_H)

# Nivel de significancia
alpha = 0.05
z_crit = norm.ppf(1 - alpha/2)

# p-valor
p_value = 2 * (1 - norm.cdf(abs(z)))

print("Z:", z)
print("z crítico:", z_crit)
print("p-valor:", p_value)

# Decisión
if abs(z) >= z_crit:
    print("Se rechaza H0")
else:
    print("No se rechaza H0")
````

## ¿Cuándo usar prueba de cola izquierda, derecha o dos colas?

La elección depende **directamente de la hipótesis alternativa \(H_1\)**, es decir, de **qué se quiere detectar**.

---

##  Prueba de dos colas

### Se usa cuando:
Se quiere detectar **cualquier cambio** (sin importar la dirección).

### Hipótesis:
$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta \neq \theta_0
$$

### Interpretación:
- Se busca si el parámetro es **distinto**
- Importa si es **mayor o menor**

### Ejemplo:
- ¿El medicamento afecta la presión?
- ¿El promedio cambió?

---

##  Cola izquierda

### Se usa cuando:
Se quiere detectar si el parámetro es **menor** que un valor.

### Hipótesis:
$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta < \theta_0
$$

### Interpretación:
- Se busca una **disminución**

### Ejemplo:
- ¿Los estudiantes tardan **menos** de 5 años?
- ¿el tiempo promedio **disminuyó**?

---

##  Cola derecha

### Se usa cuando:
Se quiere detectar si el parámetro es **mayor** que un valor.

### Hipótesis:
$$
H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta > \theta_0
$$

### Interpretación:
- Se busca un **aumento**

### Ejemplo:
- ¿Más del 40% aprueba?
- ¿el consumo **aumentó**?

---

##  Regla práctica (muy importante)

- Si el problema dice:
  - “cambió”, “diferente” → **dos colas**
  - “aumentó”, “mayor” → **cola derecha**
  - “disminuyó”, “menor” → **cola izquierda**

---

##  Error común

No se elige la cola después de ver los datos.  
La elección se hace **antes**, con base en la pregunta del problema.

---

##  Resumen corto

| Tipo de prueba | Cuándo usarla |
|------|--------|
| Dos colas | Detectar cualquier cambio |
| Cola izquierda | Detectar disminución |
| Cola derecha | Detectar aumento |

## Diferencia entre p-valor, nivel de significancia y valor crítico

###  Nivel de significancia (α)
- Umbral definido **antes del análisis**
- Representa el riesgo de **error tipo I**
- Valores comunes: 0.05, 0.01

---

###  Valor crítico
- Punto de corte en la distribución
- Depende de α y del tipo de prueba
- Define la **región de rechazo**

---

###  p-valor
- Se calcula con los datos
- Mide qué tan extremo es el resultado bajo H₀
- Indica evidencia contra H₀

---

###  Regla de decisión

- **Con valor crítico:**
  - Rechazar H₀ si el estadístico cae en la región crítica

- **Con p-valor:**
  - Rechazar H₀ si `p ≤ α`

---

###  Resumen

| Concepto       | Rol                     |
|---------------|-------------------------|
| α             | Tolerancia al error     |
| Valor crítico | Frontera de decisión    |
| p-valor       | Evidencia de los datos  |
